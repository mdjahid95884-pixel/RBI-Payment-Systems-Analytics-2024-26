```python
cols = [
    'Government_Securities_Clearing_Vol',
    'Government_Securities_Clearing_Val',
    'Forex_Clearing_Vol',
    'Forex_Clearing_Val',
    'Rupee_Derivatives_Vol',
    'Rupee_Derivatives_Val'
]

for col in cols:
    print(col)
    print(df[col].value_counts(dropna=False).head(10))
    print()
```

```python
df['GSC_Holiday'] = df['Government_Securities_Clearing_Vol'].eq('h')
```

```python
for col in cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df['BBPS_Val'] = pd.to_numeric(df['BBPS_Val'], errors='coerce')
df["CTS_Vol"] = pd.to_numeric(df['CTS_Vol'],errors='coerce')
df["CTS_Val"] = pd.to_numeric(df['CTS_Val'],errors='coerce')
```

```python
df["CreditCard_POS_Val"] = pd.to_numeric(df['CreditCard_POS_Val'],errors ="coerce")
```

```python
df['GSC_Holiday'] = df['Government_Securities_Clearing_Vol'].eq('h')
```

```python
df.describe().T
```

```python
volume_cols = [col for col in df.columns if col.endswith('_Vol')]
volume_cols
```

```python
value_cols = [col for col in df.columns if col.endswith('_Val')]
value_cols
```

```python
total_volume = df[volume_cols].sum().sort_values(ascending=False)
print("Total Transaction Count:")
print((total_volume*100000).apply(lambda x:f"⇄ {x:,.0f}"))
```

```python
daily_avg_transactions = (df[volume_cols].mean() * 100000).sort_values(ascending=False)
print(daily_avg_transactions.apply(lambda x: f"⇄ {x:,.0f}"))
```

```python
total_value = df[value_cols].sum().sort_values(ascending=False)
print("Total Transaction Amount:")
print((total_value*10000000).apply(lambda x:f"₹{x:,.0f}"))
```

```python
daily_Avg_tran_Amt = df[value_cols].mean().sort_values(ascending=False)*10000000
print(daily_Avg_tran_Amt.apply(lambda x: f"₹{x:,.0f}"))
```

```python
upi_avg_amount_per_tran = (total_value['UPI_Val'] / total_volume['UPI_Vol']) * 100
print(f"Average UPI Amount: ₹{round(upi_avg_amount_per_tran,2)}")
```

```python
rtgs_avg_amount = (total_value['RTGS_Val'] / total_volume['RTGS_Vol']) * 100
print(f"Average RTGS Amount: ₹{rtgs_avg_amount:,.0f}")
```

```python
neft_avg_amount = (total_value['NEFT_Val'] / total_volume['NEFT_Vol']) * 100
print(f"Average NEFT Amount: ₹{neft_avg_amount:,.0f}")
```

```python
imps_avg_amount = (total_value['IMPS_Val'] / total_volume['IMPS_Vol']) * 100
print(f"Average IMPS Amount: ₹{imps_avg_amount:,.0f}")
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.month_name()
```

```python
df['FY'] = df['Date'].dt.year.where(
    df['Date'].dt.month < 4,
    df['Date'].dt.year + 1
).astype(str)

df['FY'] = "FY"+((df['FY'].astype(int)-1)).astype(str)+"-"+(df['FY'].str[-2:])
```

```python
df.columns.to_list()
```

```python
upi_trend_yearly = df.groupby("FY")['UPI_Vol'].sum()*100000
print("------ UPI-TREND --------\n")
print(upi_trend_yearly.apply(lambda x:f"⇄ {x:,.0f}"))
```

```python
rtgs_trend_yearly = df.groupby("FY")['RTGS_Vol'].sum()*100000
print("------ RTGS-TREND --------\n")
print(rtgs_trend_yearly.apply(lambda x : f"{x:,.0f}"))
```

```python
imps_trend_yearly = df.groupby("FY")['IMPS_Vol'].sum()*100000
print("------ IMPS-TREND --------\n")
print(imps_trend_yearly.apply(lambda x : f"{x:,.0f}"))
```

```python
neft_trend_yearly = df.groupby("FY")['NEFT_Vol'].sum()*100000
print("------ NEFT-TREND --------\n")
print(neft_trend_yearly.apply(lambda x : f"{x:,.0f}"))
```
